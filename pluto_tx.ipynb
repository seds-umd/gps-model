```python
import adi
import numpy as np
import time
import matplotlib.pyplot as plt
```

```python
sdr = adi.Pluto('ip:192.168.20.1')
```

```python
fs = 1e6
fc = 1575.42e6
```

```python
sdr.sample_rate = int(fs)
sdr.tx_rf_bandwidth = int(fs)
sdr.tx_lo = int(fc)
sdr.tx_hardwaregain_chan0 = -50 # dB
sdr.tx_destroy_buffer()
```

```python
N = int(fs*0.05)
t = np.arange(N)/fs
x = 0.5 * np.exp(2j*np.pi*100e3*t)
x = x * 2**14

while True:
    sdr.tx(x)
```
